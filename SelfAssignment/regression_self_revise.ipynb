```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df[['mileage']], df['selling_price'], test_size=0.2, random_state=21)

```

```python
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(2)
X_train_poly = poly.fit_transform(X_train)
```

```python
from sklearn.linear_model import LinearRegression
model_poly = LinearRegression()
model_poly.fit(X_train_poly, y_train)
```

```python
x_test_poly = poly.transform(X_test)
y_pred = model_poly.predict(x_test_poly)
```

```python
from sklearn.metrics import r2_score, mean_squared_error


r2_score(y_test, y_pred), mean_squared_error(y_test, y_pred)
```

```python
from matplotlib import pyplot as plt
```

```python
X_train.shape
```

```python
x_range = np.linspace(X_train['mileage'].min(), X_train['mileage'].max(), 240).reshape(-1,1)
x_range_poly = poly.transform(x_range)
y_range_pred = model_poly.predict(x_range_poly)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(X_train['mileage'], y_train, label = 'Actual Data')
plt.plot(x_range, y_range_pred, color='orange', label='Polynomial Regression Line')
plt.title('Used Car Pricing: Polynomial Regression Fit')
plt.xlabel('Mileage')
plt.ylabel('Selling Price')
plt.legend()
plt.show()
```

### Ordinal Encoder

```python
df = pd.DataFrame({
    'Salary' : [10, 20, 30, 40, 50,],
    'age' : [1, 2, 3 , 4, 5],
    'name': ['A', 'B', 'C', 'D', 'E'],
    'gender' : ['male', 'female', 'male', 'female', 'male'],
    'location': ['kanpur', 'lucknow', 'kanpur', 'lucknow', 'kanpur'],
    'degree' : ['bsc', 'msc', 'phd', 'bsc', 'msc'],
    'level' : ['five', 'four', 'three', 'two', 'one']
})
df.head()
```

```python
replace = {'bsc' : 0, 'msc' : 1, 'phd' : 2}
df['degree_ord'] = df['degree'].map(replace)
df.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder

categories = [['bsc', 'msc', 'phd'],
              ['one', 'two', 'three', 'four', 'five']]
enc = OrdinalEncoder(categories=categories)

columns = ['degree', 'level']
encoded = enc.fit_transform(df[columns])
columns_renamed = [col_name + '_ord' for col_name in columns]
encoded_df = pd.DataFrame(encoded, columns=columns_renamed)
encoded_df
combined_df = pd.concat([df, encoded_df], axis=1)
combined_df
```

### Scaling

```python
import pandas as pd
```

```python
df = pd.read_csv(r'home_prices.csv')
df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df[['area_sqr_ft','bedrooms']], df['price_lakhs'], test_size=0.4, random_state=21)
scaler_x = MinMaxScaler().fit(X_train)
scaler_y = MinMaxScaler().fit(y_train.values.reshape(-1,1)) #reshape needed  becuase it takes 2D
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
model = LinearRegression()

model.fit(X_train,y_train)

y_pred = model.predict(X_test)

r2_score(y_test, y_pred), mean_squared_error(y_test, y_pred) #(0.844511595397575, 1077.2871677821959)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
model = LinearRegression()
X_train_scaled = scaler_x.transform(X_train)
y_train_scaled = scaler_y.transform(y_train.values.reshape(-1,1))
model.fit(X_train_scaled,y_train_scaled)

X_test_scaled = scaler_x.transform(X_test)
y_pred_scaled = model.predict(X_test_scaled)
y_pred = scaler_y.inverse_transform(y_pred_scaled)

r2_score(y_test, y_pred), mean_squared_error(y_test, y_pred) #(0.844511595397575, 1077.2871677821959)
```
