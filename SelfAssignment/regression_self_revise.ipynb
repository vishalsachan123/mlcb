```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df[['mileage']], df['selling_price'], test_size=0.2, random_state=21)

```

```python
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(2)
X_train_poly = poly.fit_transform(X_train)
```

```python
from sklearn.linear_model import LinearRegression
model_poly = LinearRegression()
model_poly.fit(X_train_poly, y_train)
```

```python
x_test_poly = poly.transform(X_test)
y_pred = model_poly.predict(x_test_poly)
```

```python
from sklearn.metrics import r2_score, mean_squared_error


r2_score(y_test, y_pred), mean_squared_error(y_test, y_pred)
```

```python
from matplotlib import pyplot as plt
```

```python
X_train.shape
```

```python
x_range = np.linspace(X_train['mileage'].min(), X_train['mileage'].max(), 240).reshape(-1,1)
x_range_poly = poly.transform(x_range)
y_range_pred = model_poly.predict(x_range_poly)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(X_train['mileage'], y_train, label = 'Actual Data')
plt.plot(x_range, y_range_pred, color='orange', label='Polynomial Regression Line')
plt.title('Used Car Pricing: Polynomial Regression Fit')
plt.xlabel('Mileage')
plt.ylabel('Selling Price')
plt.legend()
plt.show()
```

```python
df = pd.read_csv(r'mpg.csv')
df.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder
enc = OrdinalEncoder()
enc.fit(df[['car_name']])
```

```python
df['ordinal_car_name'] = enc.transform(df[['car_name']])
df.head()
```

Now your model might think:

Ford (2) > BMW (1) > Maruti (0)

as if there is some order or distance between them!

But that’s wrong ❌ — these are just names, not ordered values.
(“BMW” isn’t twice “Maruti”)

So — we need a way to represent categories without implying order.

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.DataFrame({
    'car_name': ['Maruti', 'BMW', 'Ford', 'BMW', 'Maruti'],
    'gender': ['Male', 'Female','Male', 'Female','Male'],
    'selling_price': [300000, 1200000, 800000, 1100000, 320000]
})
df.head()
```

```python
enc = OneHotEncoder()
encoded = enc.fit_transform(df[['car_name']]).toarray()
col_names = enc.get_feature_names_out(['car_name'])
print(col_names)
```

```python
encoded_df = pd.DataFrame(encoded, columns=col_names)
encoded_df.head()
```

```python
df_final = pd.concat([df.drop('car_name', axis=1), encoded_df], axis=1)
df_final.head()
```

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    'education': ['BSc', 'MSc', 'BSc', 'PhD', 'MSc'],
})

# Define your intentional order
edu_order = [['BSc', 'MSc', 'PhD']]

# Create encoder with custom order
enc = OrdinalEncoder(categories=edu_order)

# Fit + transform
df['edu_encoded'] = enc.fit_transform(df[['education']])

print(df)
```

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    'education': ['BSc', 'MSc', 'BSc', 'PhD', 'MSc'],
    'education2': ['BSc', 'MSc', 'BSc', 'PhD', 'MSc']
})

# Define your intentional order
edu_order = [
    ['BSc', 'MSc', 'PhD'],   # for 'education'
    ['BSc', 'MSc', 'PhD']    # for 'education2'
]
# Create encoder with custom order
enc = OrdinalEncoder(categories=edu_order)
df[['edu_encoded1', 'edu_encoded2']] = enc.fit_transform(df[['education', 'education2']])

print(df)
```
