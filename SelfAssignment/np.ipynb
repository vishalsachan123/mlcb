```python
import numpy as np
import matplotlib.pyplot as plt

# Generate angles from 0 to 4π radians
theta_values = np.linspace(0, 4*np.pi, 1000)
sin_values = np.sin(theta_values)
cos_values = np.cos(theta_values)

# Calculate tangent, but mask values near asymptotes
tan_values = np.tan(theta_values)
# Clip tangent values for plotting (avoid extreme values near ±∞)
tan_values = np.clip(tan_values, -10, 10)  # Limit y-axis for visibility

# Create a figure with two subplots
plt.figure(figsize=(12, 5))

# Subplot 1: Unit Circle (cos(θ), sin(θ))
plt.subplot(1, 2, 1)  # 1 row, 2 columns, 1st subplot
plt.scatter(cos_values, sin_values, color="blue", s=10, label="Unit Circle")
plt.xlabel("cos(θ)")
plt.ylabel("sin(θ)")
plt.title("Unit Circle: cos(θ) vs sin(θ)")
plt.grid(True)
plt.axis("equal")  # Ensure circle isn’t distorted
plt.legend()

# Subplot 2: Tangent Function (θ vs tan(θ))
plt.subplot(1, 2, 2)  # 1 row, 2 columns, 2nd subplot
plt.plot(theta_values, tan_values, color="green", label="tan(θ)")
plt.xlabel("θ (radians)")
plt.ylabel("tan(θ)")
plt.title("Tangent Function")
plt.grid(True)
plt.ylim(-10, 10)  # Limit y-axis to avoid extreme values
plt.legend()

# Adjust layout to prevent overlap
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate angles for the unit circle (0 to 2π)
theta_values = np.linspace(0, 2*np.pi, 1000)
cos_values = np.cos(theta_values)
sin_values = np.sin(theta_values)

# Specific angle for the line (e.g., 45° = π/4 radians)
theta = np.radians(45)  # Convert 45° to radians
x_point = np.cos(theta)  # x = cos(θ)
y_point = np.sin(theta)  # y = sin(θ)
tan_value = np.tan(theta)  # tan(θ) = sin(θ)/cos(θ)

# Create the plot
plt.figure(figsize=(8, 6))

# Plot the unit circle
plt.plot(cos_values, sin_values, color="blue", label="Unit Circle")

# Plot the line from origin to point on circle
plt.plot([0, x_point], [0, y_point], color="red", label=f"Line at θ=45°")

# Optionally extend to x=1 to show tan(θ)
plt.plot([0, 1], [0, tan_value], color="green", linestyle="--", label=f"tan(45°) = {tan_value:.2f}")

# Plot the vertical line x=1 for tangent visualization
plt.axvline(x=1, color="gray", linestyle="--", label="x=1 line")

# Mark the point on the circle
plt.scatter([x_point], [y_point], color="red", s=100, label="Point (cos(θ), sin(θ))")

# Mark the tangent point on x=1
plt.scatter([1], [tan_value], color="green", s=100, label="Tangent point")

plt.xlabel("x (cos(θ))")
plt.ylabel("y (sin(θ))")
plt.title("Unit Circle with Line and Tangent")
plt.grid(True)
plt.axis("equal")  # Ensure circle is not distorted
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, uniform, expon

# Generate x values
x = np.linspace(-5, 5, 1000)

# Distributions
normal_pdf = norm.pdf(x, loc=0, scale=1)  # Normal: μ=0, σ=1
uniform_pdf = uniform.pdf(x, loc=-2, scale=4)  # Uniform: [-2, 2]
exponential_pdf = expon.pdf(x, scale=1)  # Exponential: λ=1

# Plot
plt.figure(figsize=(10, 6))
plt.plot(x, normal_pdf, label="Normal (μ=0, σ=1)", color="blue")
plt.plot(x, uniform_pdf, label="Uniform [-2, 2]", color="red")
plt.plot(x, exponential_pdf, label="Exponential (λ=1)", color="green")
plt.xlabel("x")
plt.ylabel("Probability Density")
plt.title("Common ML Distributions")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import expon

# समय (0 से 1 मिनट, क्योंकि scale=0.1 के लिए छोटा समय उपयुक्त)
x = np.linspace(0, 1, 1000)

# PDFs अलग-अलग scale के साथ
exp_pdf_0_1 = expon.pdf(x, scale=0.1)  # औसत 0.1 मिनट (λ=10)
exp_pdf_1 = expon.pdf(x, scale=1)      # औसत 1 मिनट (λ=1)
exp_pdf_100 = expon.pdf(x, scale=100)  # औसत 100 मिनट (λ=0.01)

# प्लॉट
plt.figure(figsize=(8, 6))
plt.plot(x, exp_pdf_0_1, label="औसत 0.1 मिनट (scale=0.1, λ=10)", color="blue")
plt.plot(x, exp_pdf_1, label="औसत 1 मिनट (scale=1, λ=1)", color="red")
plt.plot(x, exp_pdf_100, label="औसत 100 मिनट (scale=100, λ=0.01)", color="green")
plt.xlabel("समय (मिनट)")
plt.ylabel("प्रायिकता घनत्व")
plt.title("Exponential Distribution: scale=0.1 vs अन्य")
plt.grid(True)
plt.legend()
plt.show()
```

```python
from scipy.stats import poisson

k = np.arange(0, 15)
poisson_pmf = poisson.pmf(k, mu=5)  # औसत 5 ग्राहक
poisson_pmf_2 = poisson.pmf(k, mu=2)  # औसत 2 ग्राहक

plt.figure(figsize=(8, 6))
plt.stem(k, poisson_pmf, label="औसत=5 ग्राहक", linefmt="blue", markerfmt="bo")
plt.stem(k, poisson_pmf_2, label="औसत=2 ग्राहक", linefmt="red", markerfmt="ro")
plt.xlabel("ग्राहकों की संख्या")
plt.ylabel("प्रायिकता")
plt.title("Poisson Distribution: दुकान में ग्राहक")
plt.grid(True)
plt.legend()
plt.show()
```

```python
from scipy.stats import bernoulli

k = [0, 1]
bernoulli_pmf = bernoulli.pmf(k, p=0.7)  # p=0.7
bernoulli_pmf_2 = bernoulli.pmf(k, p=0.3)  # p=0.3

plt.figure(figsize=(8, 6))
plt.bar([0, 1], bernoulli_pmf, width=0.4, label="p=0.7 (खरीदने की संभावना)", color="blue")
plt.bar([0.4, 1.4], bernoulli_pmf_2, width=0.4, label="p=0.3", color="red")
plt.xticks([0, 0.4, 1, 1.4], ["नहीं (0)", "नहीं (0)", "हाँ (1)", "हाँ (1)"])
plt.ylabel("प्रायिकता")
plt.title("Bernoulli Distribution: ग्राहक का खरीदना")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
from scipy.stats import expon
#suppose you get 2 customers per minutes
# how likely it is to get a customers at time which will decrese as time increse # use cdf to get area under height
lam = 2 #avg num of events per unit time
minutes = [0.1, 0.2,0.3,0.4,1,2,3]
y = expon.pdf(minutes, loc=0, scale=1/lam)
y# ([1.63746151, 1.34064009, 1.09762327, 0.89865793, 0.27067057, 0.03663128, 0.0049575 ])
```

```python
import numpy as np
from scipy.stats import expon

lam = 2
times = [0.1, 0.3, 0.5, 1, 2,20]
pdf = expon.pdf(times, scale=1/lam)
cdf = expon.cdf(times, scale=1/lam) # within that time

for t, p, c in zip(times, pdf, cdf):
    print(f"Time={t:>3} min → PDF={p:>6.3f},   CDF={c:>6.3f}")
```
