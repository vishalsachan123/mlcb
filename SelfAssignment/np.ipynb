```python
import numpy as np
```

```python
arr = np.array(['abc', 'defg'], dtype='U')
print(arr)
print(arr[0])
print(type(arr))
print(type(arr[0]))
print(arr.dtype)
```

```python
arr = np.array(['abc', 'defg'], dtype='S')
print(arr)
print(arr[0])
print(type(arr))
print(type(arr[0]))
print(arr.dtype)
print(arr[0].decode('utf'))
```

```python

arr_S = np.array(['a', 'b', 'c'], dtype='S')
arr_U = np.array(['a', 'b', 'c'], dtype='U')

print(arr_S.nbytes)   # smaller 3
print(arr_U.nbytes)   # larger 12
```

```python
arr = np.arange(2*3*4*5*3).reshape(2, 3, 4, 5, 3)
print(arr.shape)
# print(arr)

print(arr[0].shape)        # (3, 4, 5, 3)
print(arr[0, 1].shape )    # (4, 5, 3)
print(arr[0, 1, 2].shape ) # (5, 3)
print(arr[0, 1, 2, 4].shape)  # (3,)

```
