```python

x = np.linspace(-5, 5, 500)
y = x**3

plt.figure(figsize=(8, 6))
plt.plot(x, y, label="y = x²", color="blue")
plt.title("Graph of y = x²")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.axhline(0, color="black", linewidth=0.5)
plt.axvline(0, color="black", linewidth=0.5)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3, 3, 100)
y1 = x**3
y2 = x**3 + x**2 + 2*x + 1

plt.plot(x, y1, label="y = x³", linestyle='--')
plt.plot(x, y2, label="y = x³ + x² + 2x + 1", color='red')
plt.legend()
plt.grid(True)
plt.title("Comparing Polynomial Equations")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Create x values from -2 to 2
x = np.linspace(-2, 2, 400)

# Calculate y values for x^3 and x^5
y1 = x**3
y2 = x**5 #steep slope

# Plot the graphs
plt.figure(figsize=(8,5))
plt.plot(x, y1, label='y = x^3', color='blue')
plt.plot(x, y2, label='y = x^5', color='red')
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.title('Comparison of y = x^3 and y = x^5')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```
