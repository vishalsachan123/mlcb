```python
df = pd.read_csv(r'datasets\batter.csv')
df = df.sample(50)
df
```

```python
plt.scatter(df['avg'], df['strike_rate'], color='red', marker='+', s = 50)
plt.title('Avg and Strik Rate')
plt.xlabel('avg')
plt.ylabel('strike_rate')
```

```python
type(df[['strike_rate']])#pandas.core.frame.DataFrame
type(df['strike_rate'])#pandas.core.series.Series
```

```python
tips = sns.load_dataset('tips')
tips = tips.head(50)
tips
```

```python
plt.scatter(tips['total_bill'], tips['tip'])
plt.title('Tips data')
plt.xlabel('total_bill')
plt.ylabel('tip')
```

```python
plt.scatter(tips['total_bill'], tips['tip'],s = tips['size']*20)
plt.title('Tips data') 
plt.xlabel('total_bill')
plt.ylabel('tip')
```

```python
plt.plot(tips['total_bill'], tips['tip'], 'o')#this is fast
plt.title('Tips data')
plt.xlabel('total_bill') 
plt.ylabel('tip')
```
