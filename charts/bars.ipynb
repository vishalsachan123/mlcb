```python
#stack bar
plt.bar(df['batsman'], df['2015'],label='2015')
plt.bar(df['batsman'], df['2016'],label='2016',bottom=df['2015'])
plt.bar(df['batsman'], df['2017'],label='2017',bottom=(df['2015'] + df['2016']))
plt.legend()
plt.show()
```
