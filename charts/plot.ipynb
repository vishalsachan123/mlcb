```python
batsman = pd.read_csv(r'datasets\sharma-kohli.csv')
batsman
```

```python
plt.plot(batsman['index'], batsman['V Kohli'], label = 'V kohli')
plt.plot(batsman['index'], batsman['RG Sharma'], label = 'RG Sharma')

plt.title('Virat Vs Rohit')
plt.legend()
plt.xlabel('Season')
plt.ylabel('Runs')
```

```python
plt.plot(batsman['index'], batsman['V Kohli'], color = 'blue', linestyle='dashed', linewidth = 3)
plt.plot(batsman['index'], batsman['RG Sharma'], color = 'red', linestyle='dotted')#dashdot

plt.title('Virat Vs Rohit')
plt.xlabel('Season')
plt.ylabel('Runs')
```

```python
plt.plot(batsman['index'], batsman['V Kohli'], color = 'blue', linestyle='dashed', linewidth = 3, marker= 'D', markersize = 10)
plt.plot(batsman['index'], batsman['RG Sharma'], color = 'red', linestyle='dotted',marker= 'o')#dashdot

plt.title('Virat Vs Rohit')
plt.xlabel('Season')
plt.ylabel('Runs')
```

```python
plt.plot(batsman['index'], batsman['V Kohli'], color = 'blue', linestyle='dashed', linewidth = 3, marker= 'D', markersize = 10, label = 'Virat')
plt.plot(batsman['index'], batsman['RG Sharma'], color = 'red', linestyle='dotted',marker= 'o', label = 'Rohit')#dashdot

plt.title('Virat Vs Rohit')
plt.xlabel('Season')
plt.ylabel('Runs')
plt.legend() # to show labels
plt.grid()# to activat grid
#plt.show() to show sometimes in others
```

```python
plt.plot(batsman['index'], batsman['RG Sharma'],linestyle='dashed', linewidth=2, color='blue', marker='o', label = 'RG Sharma')
plt.plot(batsman['index'], batsman['V Kohli'], linestyle='dashed', linewidth=1, color = 'red', marker='+', label = 'V Kohli')
plt.grid()
plt.legend()
plt.xlim(2010, 2014)
```
