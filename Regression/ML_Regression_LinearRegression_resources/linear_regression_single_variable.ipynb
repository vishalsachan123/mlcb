<h2 align="center"> Linear Regression Single Variable Tutorial </h2>

Predict home price using area per square ft

```python
import pandas as pd

df = pd.read_csv("home_prices.csv")
df.head()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df['area_sqr_ft'], df['price_lakhs'])
plt.xlabel('Square Footage')
plt.ylabel('Price')
plt.title('Home Prices vs. Square Footage')
plt.show()
```

now the model will try to find a best fit line

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(df[['area_sqr_ft']], df['price_lakhs'])# it takes 2D array
```

```python
model.predict([[1500],[700]])
```

```python
model.coef_, model.intercept_
```

```python
model.coef_[0]*1550 + model.intercept_
```

```python
model.coef_[0]*700 + model.intercept_
```
